```python
import pandas as pd
from sqlalchemy import create_engine
import os

file_name = 'Dataset for Data Analytics3.xlsx'

if os.path.exists(file_name):
    df = pd.read_excel(file_name)
    print(" Success: Excel File found!")
    
    engine = create_engine('sqlite:///project3_final_db.db')
    
    df.to_sql('project_data', engine, index=False, if_exists='replace')
    print("✅ Success: SQL Table 'project_data' mei!")
    
    query = """
    SELECT 
        Product, 
        COUNT(*) as Total_Orders, 
        SUM(Quantity) as Total_Units_Sold,
        SUM(TotalPrice) as Total_Revenue
    FROM project_data
    GROUP BY Product
    ORDER BY Total_Revenue DESC;
    """
    
    result = pd.read_sql(query, engine)
    print("\n--- Project 3: Final SQL Output ---")
    display(result)

else:
    print(f"❌ Error: '{file_name}' nahi mili.")
    print("Aapke Home folder mein ye files mojood hain:")
    print(os.listdir())
```

```python
# Is query ko aik naye cell mein run karein
query_filter = "SELECT * FROM project_data WHERE OrderStatus = 'Delivered' LIMIT 10;"
df_filter = pd.read_sql(query_filter, engine)
df_filter
```

```python
# Naye cell mein ye run karein
query_where = """
SELECT OrderID, CustomerID, Product, TotalPrice, OrderStatus
FROM project_data
WHERE OrderStatus = 'Cancelled' AND TotalPrice > 1000
ORDER BY TotalPrice DESC;
"""
df_cancelled = pd.read_sql(query_where, engine)
print("Task: High-Value Cancelled Orders")
df_cancelled
```

```python
# Ek aur naye cell mein ye run karein
query_avg = """
SELECT 
    Product, 
    COUNT(*) as Total_Sales_Count,
    AVG(UnitPrice) as Average_Unit_Price
FROM project_data
GROUP BY Product
ORDER BY Average_Unit_Price DESC;
"""
df_avg = pd.read_sql(query_avg, engine)
print("Task: Product Price Analysis (Averages)")
df_avg
```
